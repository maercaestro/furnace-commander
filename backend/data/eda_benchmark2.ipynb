# EDA for Benchmark Data Part 2
This time, we will specifically look at the data dynamics and determine whether we have enough data dyanmics for the model to learn. If not, we may need to train PINN

```python
#first, let's load our smooth data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('smoothed_data.csv')
df.info()
```

```python
#the data is clean, let's try to find the std for each featueres
feature_cols = [col for col in df.columns if col != 'timestep']
diff_std = df[feature_cols].diff().std()
print("Std of first differences per feature:\n",diff_std)
```

```python
#let's print unique value per features
print("Unique values (per feature):\n", df[feature_cols].nunique())
```

```python
#let's print the range
print("Range (per feature):\n", df[feature_cols].max() - df[feature_cols].min())
```

 Based on here, we see that the data has low dynamics between each other causing the network to not able to learn well and generalize well given different conditions.
